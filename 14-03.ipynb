```python
df.acceleration.describe()
```

```python
q1=df.acceleration.quantile(0.25)
q1
```

```python
q3=df.acceleration.quantile(0.75)
q3
```

```python
iqr=q3-q1
iqr
```

```python
ll=q1-1.5*iqr
hl=q3+1.5*iqr
```

```python
ll
```

```python
hl
```

```python
df[(df["acceleration"]>hl)|(df["acceleration"]<ll)]
```

```python
df=df[(df["acceleration"]<=hl)&(df["acceleration"]>=ll)]
```

```python
df
```

```python
sns.boxplot(data=df,y="acceleration")
```
