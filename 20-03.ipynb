```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.io as pio
pio.renderers.default="notebook"
```

```python
df=pd.read_csv("social_media_engagement.csv")
```

```python
df
```

```python
px.bar(data_frame=df,x="post_type",y="likes")
```

```python
px.bar(data_frame=df,x="post_type",y="likes",height=1000,width=700,title="post")
```

```python
px.bar(data_frame=df,x="post_type",y="likes",height=1000,width=700,title="post",hover_data=["comments","shares","post_day"])
```

```python
px.bar(data_frame=df,x="post_type",y="likes",height=1000,width=700,
       title="post",hover_data=["comments","shares","post_day"],text='likes')
```

```python
px.bar(data_frame=df,x="post_type",y="likes",height=1000,width=700,
       title="post",hover_data=["comments","shares","post_day"],text_auto=True)
```

```python
px.bar(data_frame=df,x="post_type",y="likes",height=1000,width=700,
       title="post",hover_data=["comments","shares","post_day"],
       text_auto=True,color="platform",color_discrete_sequence=["red","green","blue"])
```

```python
px.bar(data_frame=df,x="post_type",y="likes",height=1000,width=700,
       title="post",hover_data=["comments","shares","post_day"],
       text_auto=True,color="comments")
```

```python
px.bar(data_frame=df,x="post_type",y="likes",height=1000,width=700,
       title="post",hover_data=["comments","shares","post_day"],
       text_auto=True,color="platform",pattern_shape="sentiment_score")
```

```python
px.bar(data_frame=df,x="post_type",y="likes",height=1000,width=700,
       title="post",hover_data=["comments","shares","post_day"],
       text_auto=True,color="platform",pattern_shape="sentiment_score",
       facet_row="sentiment_score",facet_col="platform")
```

```python
px.bar(data_frame=df,x="post_type",y=["likes","comments","shares"])
```

```python
px.bar(data_frame=df,x="post_type",y=["likes","comments","shares"],barmode="stack")
```

```python
px.bar(data_frame=df,x="post_type",y=["likes","comments","shares"],barmode="group")
```
