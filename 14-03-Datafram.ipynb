```python
import pandas as pd
import numpy as np
```

```python
df=pd.DataFrame(np.random.randint(15,100,(20,4)),columns=["A","B","C","D"])
df
```

```python
df["Avg"]=df.apply("mean",axis=1)
```

```python
df.apply(lambda x:x["A"]+x["B"]+x["C"],axis=1)
```

```python
df["Grade"]=df.apply(lambda x:"Fail" if x["A"]<35 or x["B"]<35 or x["C"]<35 or x["D"]<35 else "pass",axis=1)
```

```python
def result(x):
    if x>90:
        return "A"
    if x>80:
        return "B"
    if x>70:
        return "C"
    if x>60:
        return "D"
    return "FAIL"
```

```python
df["Avg"].apply(result)
```
